# Занятие 9. Функция активации и функция потерь

**Демонстрация и практика · модуль 4 «Нейронные сети»**

**Главная идея:** Активация отвечает «какое значение пойдёт дальше?», а loss — «насколько неудачен ответ относительно цели?».

Запускайте ячейки сверху вниз. Задания для самостоятельной работы расположены в конце.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
plt.rcParams.update({'figure.figsize': (8, 4), 'axes.grid': True, 'grid.alpha': 0.2})
```

## 1. Две функции активации

```python
z = np.linspace(-6,6,400)
def sigmoid(x): return 1/(1+np.exp(-x))
def relu(x): return np.maximum(x,0)
fig, axes=plt.subplots(1,2,figsize=(11,4))
axes[0].plot(z,sigmoid(z));axes[0].set_title('Sigmoid: плавный выход 0..1')
axes[1].plot(z,relu(z));axes[1].set_title('ReLU: отрицательные превращаются в 0')
plt.show()
print('z=-2,0,2:',list(zip(sigmoid(np.array([-2.,0,2])).round(3),relu(np.array([-2.,0,2])))))
```

## 2. Loss — отдельная функция
Модели могут давать одинаковый итоговый класс, но разные потери.

```python
def mse(p,y): return (p-y)**2
def bce(p,y):
    p=np.clip(p,1e-7,1-1e-7)
    return -(y*np.log(p)+(1-y)*np.log(1-p))
for p in [0.2,0.51,0.9,0.99]:
    print(f'y=1, p={p}: MSE={mse(p,1):.3f}, BCE={bce(p,1):.3f}, класс={int(p>=0.5)}')
```

## 3. График штрафа

```python
p=np.linspace(.001,.999,300)
plt.plot(p,bce(p,1),label='если правильный класс 1')
plt.plot(p,bce(p,0),label='если правильный класс 0')
plt.xlabel('выход p');plt.ylabel('BCE loss');plt.title('Уверенная ошибка стоит дорого');plt.legend();plt.show()
```
