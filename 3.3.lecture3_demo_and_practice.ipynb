## Часть 1. Демонстрация преподавателя

Цель — проследить полный путь изображения: **файл → Pillow → NumPy → Matplotlib** и научиться проверять каждый этап.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image
import sklearn.datasets

plt.rcParams['figure.figsize'] = (8, 4)

# Берём JPEG, который уже входит в scikit-learn: интернет не нужен.
sample_path = Path(sklearn.datasets.__file__).resolve().parent / 'images' / 'china.jpg'
print('Файл:', sample_path)
print('Существует:', sample_path.exists())
```

### 1. Открываем именно файл
`Image.open` читает формат изображения и возвращает объект Pillow. Это ещё не NumPy-массив.

```python
pil_image = Image.open(sample_path)
print('type:', type(pil_image))
print('format:', pil_image.format)
print('mode:', pil_image.mode)
print('size (W, H):', pil_image.size)
```

### 2. Переводим изображение в NumPy
Для численных операций приводим режим к RGB и получаем массив.

```python
pil_rgb = pil_image.convert('RGB')
image = np.array(pil_rgb)
print('shape (H, W, C):', image.shape)
print('dtype:', image.dtype)
print('min/max:', image.min(), image.max())
```

### 3. Показываем RGB-массив
`imshow` получает уже готовый массив и визуализирует его.

```python
plt.imshow(image)
plt.title('RGB-массив')
plt.axis('off')
plt.show()
```

### 4. 2D-массив нужно визуализировать осознанно
Сделаем серую версию. Без `cmap='gray'` Matplotlib раскрасит числа своей цветовой картой.

```python
gray = np.array(pil_rgb.convert('L'))
fig, axes = plt.subplots(1, 2, figsize=(9, 4))
axes[0].imshow(gray); axes[0].set_title('без cmap')
axes[1].imshow(gray, cmap='gray', vmin=0, vmax=255); axes[1].set_title("cmap='gray'")
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```

### 5. Координаты: `image[y, x]`, но на графике точка `(x, y)`
Отметим точку и распечатаем RGB-вектор в этой позиции.

```python
y, x = 220, 360
print('image[y, x] =', image[y, x])
plt.imshow(image)
plt.scatter([x], [y], s=80)
plt.title(f'точка (x={x}, y={y})')
plt.show()
```

### 6. Срез массива как увеличение области
Срез выбирает прямоугольную область. Никакого resize здесь нет.

```python
crop = image[140:320, 250:520]
print('Исходный shape:', image.shape, 'crop:', crop.shape)
plt.imshow(crop)
plt.axis('off')
plt.show()
```

### 7. Несколько представлений рядом
Так проще сравнивать исходник, выбранную область и серую версию.

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(image); axes[0].set_title('исходник')
axes[1].imshow(crop); axes[1].set_title('crop')
axes[2].imshow(gray, cmap='gray', vmin=0, vmax=255); axes[2].set_title('gray')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```
