# Занятие 8. Линейный классификатор и искусственный нейрон

**Демонстрация и практика · модуль 4 «Нейронные сети»**

**Главная идея:** Один нейрон не видит изображение как человек: он получает числа, умножает их на веса и получает оценку.

Запускайте ячейки сверху вниз. Задания для самостоятельной работы расположены в конце.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
plt.rcParams.update({'figure.figsize': (8, 4), 'axes.grid': True, 'grid.alpha': 0.2})
```

## 1. Взвешенная сумма и порог
Сначала смотрим как нейрон считает.

```python
x = np.array([1, 0])
w = np.array([2., -1.])
b = -0.5
z = np.dot(x, w) + b
print('вход:', x, '| z =', z, '| класс =', int(z > 0))
```

## 2. Пороговые нейроны AND и OR
Одни входы — разные смещения.

```python
examples = np.array([[0,0],[0,1],[1,0],[1,1]])
def threshold_neuron(x, w, b):
    return (np.asarray(x) @ np.asarray(w) + b > 0).astype(int)
for name, bias in [('AND', -1.5), ('OR', -0.5)]:
    print(name, '→', threshold_neuron(examples, [1,1], bias))
```

## 3. Граница решения на плоскости
Меняем веса — меняем ориентацию границы; меняем b — сдвигаем.

```python
fig, ax = plt.subplots()
for pt in examples:
    ax.scatter(*pt, c='tab:blue' if threshold_neuron(pt,[1,1],-1.5) else 'tab:orange', s=110)
t = np.linspace(-.4, 1.4, 100)
for bias, style in [(-1.5, '-'),(-0.5, '--')]:
    ax.plot(t, -t-bias, style, label=f'b={bias}')
ax.set(xlim=(-.3,1.4),ylim=(-.3,1.4),xlabel='x₁',ylabel='x₂',title='Граница w₁x₁+w₂x₂+b=0')
ax.legend();plt.show()
```
