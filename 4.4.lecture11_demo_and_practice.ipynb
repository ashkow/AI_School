## 1. Четыре точки XOR
Класс 1 означает «включён ровно один переключатель». Попробуйте отделить оранжевые точки одной прямой. Затем проверьте правило «между двумя параллельными прямыми».

```python
import numpy as np
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 3.8), layout='constrained')
for label, color in [(0, '#245c8b'), (1, '#c75925')]:
    ax.scatter(*X[y == label].T, s=145, color=color, label=f'Класс {label}', zorder=3)
grid = np.linspace(-0.25, 1.25, 100)
ax.fill_between(grid, 0.5-grid, 1.5-grid, alpha=0.16, color='#c75925')
ax.set(xlim=(-0.25, 1.25), ylim=(-0.25, 1.25), xticks=[0, 1], yticks=[0, 1],
       xlabel='x₁', ylabel='x₂', title='Две границы выделяют полосу')
ax.set_aspect('equal')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.23), ncol=2, frameon=False)
plt.show()
```

## 2. Скрытый слой
Пусть `s = x1 + x2`. Первый нейрон пропускает сумму, второй с помощью ReLU пропускает только её часть выше 1. Выходной нейрон считает `h1 - 2*h2`.

```python
def relu(z):
    return np.maximum(z, 0)

def hidden_layer(x):
    total = x[:, 0] + x[:, 1]
    return np.column_stack([relu(total), relu(total - 1)])

hidden = hidden_layer(X)
scores = hidden[:, 0] - 2 * hidden[:, 1]
print('x1 x2 | h1 h2 | выход | XOR')
for pair, hints, score, label in zip(X, hidden, scores, y):
    print(*pair, '|', *hints, '|', score, '|', label)
```

## 3. Эксперимент без ReLU
Вообразите, что оба нейрона пропускают любое число. Для входа (0, 0) второй нейрон выдаст −1. Как изменится ответ?

```python
total = X.sum(axis=1)
linear_hidden = np.column_stack([total, total - 1])
linear_scores = linear_hidden[:, 0] - 2 * linear_hidden[:, 1]
print('С ReLU:', scores)
print('Без ReLU:', linear_scores)
print('Та же последовательность из формулы 2 − s:', 2 - total)
```
